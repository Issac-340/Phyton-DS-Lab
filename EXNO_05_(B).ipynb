```python
class Evaluate:

    # Constructor to initialize the class
    def __init__(self, capacity):
        self.top = -1
        self.capacity = capacity
        # This array is used as a stack
        self.array = []

    # Check if the stack is empty
    def isEmpty(self):
        return True if self.top == -1 else False

    # Return the top element of the stack
    def peek(self):
        if not self.isEmpty():
            return self.array[-1]
        return None # or raise an error for empty stack

    # Pop an element from the stack
    def pop(self):
        if not self.isEmpty():
            self.top -= 1
            return self.array.pop()
        else:
            # A common practice is to raise an error for popping from an empty stack
            # or return a specific indicator like None, not just '$'
            return None

    # Push an element onto the stack
    def push(self, op):
        self.top += 1
        self.array.append(op)

    # Evaluate the postfix expression
    def evaluatePostfix(self, exp):
        for i in exp:
            if i.isdigit():
                self.push(i)
            else:
                # Ensure there are enough operands for the operation
                val1 = self.pop()
                val2 = self.pop()

                # Handle potential None returns from pop for empty stack
                if val1 is None or val2 is None:
                    raise ValueError("Invalid postfix expression: Not enough operands for operator '" + i + "'")

                # Convert values to int for calculation, then back to string for stack
                self.push(str(eval(val2 + i + val1)))

        # The final result should be the only element left on the stack
        result = self.pop()
        if result is None or not self.isEmpty():
            raise ValueError("Invalid postfix expression: Malformed expression")

        return int(result)

# Example usage:
exp = "231*+9-"
obj = Evaluate(len(exp))

# Handle potential errors during evaluation
try:
    print("postfix evaluation: %d" % (obj.evaluatePostfix(exp)))
except ValueError as e:
    print(f"Error: {e}")
```
