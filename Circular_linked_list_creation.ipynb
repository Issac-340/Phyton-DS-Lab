```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CreateList:
    def __init__(self):
        self.head = None
        self.tail = None

    def add(self, data):
        newNode = Node(data)
        if self.head is None:
            self.head = newNode
            self.tail = newNode
            newNode.next = self.head # For a single-node circular list, points to itself
        else:
            self.tail.next = newNode
            self.tail = newNode
            self.tail.next = self.head # Ensure the new tail points back to head

    def display(self):
        current = self.head
        if self.head is None:
            print("List is empty")
            return
        else:
            print("Nodes of the circular linked list: ")
            print(current.data) # Print data of the first node
            current = current.next # Move to the next node
            while current != self.head: # Iterate until we circle back to the head
                print(current.data)
                current = current.next

# Example usage
cl = CreateList()
cl.add(1)
cl.add(2)
cl.add(3)
cl.add(4)
cl.display()
```
