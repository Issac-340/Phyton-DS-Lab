```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLinkedList:
    def __init__(self):
        self.last = None

    def addToEmpty(self, data):
        if self.last is not None:
            return self.last # If list is not empty, return existing last (or handle error)

        temp = Node(data)
        self.last = temp
        self.last.next = self.last
        return self.last

    def addBegin(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next # New node points to current head
        self.last.next = temp # Last node now points to new node
        return self.last

    def addEnd(self, data):
        if self.last is None:
            return self.addToEmpty(data)

        temp = Node(data)
        temp.next = self.last.next # New node points to current head
        self.last.next = temp # Last node points to new node
        self.last = temp # New node becomes the last node

        return self.last

    def addAfter(self, data, item):
        if self.last is None:
            print("List is empty, cannot add after an item.")
            return None

        temp = Node(data)
        p = self.last.next # Start from the head
        found = False

        while True:
            if p.data == item:
                temp.next = p.next
                p.next = temp
                if p == self.last: # If added after the last node, the new node becomes the last
                    self.last = temp
                found = True
                break
            p = p.next
            if p == self.last.next: # Traversed the entire list without finding the item
                break

        if not found:
            print(f"{item} not present in the list")
        return self.last

    def traverse(self):
        if self.last is None:
            print("List is empty")
            return

        temp = self.last.next # Start from the head
        while True:
            print(temp.data, end=" ")
            temp = temp.next
            if temp == self.last.next: # Stop when we return to the head
                break
        print() # Newline after printing all elements

if __name__ == '__main__':
    llist = CircularLinkedList()

    # Operations
    llist.addToEmpty(6)
    llist.addBegin(4)
    llist.addBegin(2)
    llist.addEnd(8)
    llist.addEnd(12)
    llist.addAfter(10, 8)

    print("Circular Linked List:")
    llist.traverse()

    llist.addAfter(100, 50) # Test case for item not found
    llist.traverse()

    llist.addAfter(1, 12) # Add after the last element (12)
    llist.traverse()
```
